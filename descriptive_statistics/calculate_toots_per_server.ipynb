```python
import sys
import glob
import os
from os.path import join
from tqdm import tqdm
sys.path.append(os.path.dirname(os.path.abspath('.')))
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

from config import DATA_DIR
```

```python
toots_per_instance = {}

for filepath in tqdm(glob.glob(join(DATA_DIR, "05-timeline_raw", "*.jsonl"))):
    instance_id = os.path.basename(filepath).replace(".jsonl", "")
    count = 0
    with open(filepath, "r") as f:
        for line in f:
            if line.strip():
                count += 1
    toots_per_instance[instance_id] = count

toots_per_instance_df = pd.Series(toots_per_instance, name="toots_per_server")
print(toots_per_instance_df.describe())
```

```python
toots_per_instance_df = toots_per_instance_df.to_frame()
toots_per_instance_df["has_toots"] = (toots_per_instance_df["toots_per_server"] > 0).astype(int)
```

```python
toots_per_instance_df.to_parquet(join(DATA_DIR, "toots_per_server_has_toots.parquet"))
```
