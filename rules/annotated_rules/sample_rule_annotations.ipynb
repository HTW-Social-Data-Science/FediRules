# Cohen's Kappa

```python
# load file after manual annotation by annotator 2
df_anno_2 = pd.read_csv('sample_annotated_rules_annotator_2.csv', sep = ",")
df_anno_2.head()
```

### clear wrong row and summarize cw-rows

```python
sample_df = sample_df[sample_df["translated_rules"] != "nudity_nocw pornography_nocw illegalContentLinks spam advertising spoilers_nocw"]
df_anno_2 = df_anno_2[df_anno_2["translated_rules"] != "nudity_nocw pornography_nocw illegalContentLinks spam advertising spoilers_nocw"]
```

```python
cw_columns = [
    'porn_nocw',
    'violent_no_cw',
    'nsfw_no_cw',
    'spoilers_no_cw',
    'sensitive_no_cw',
    'politics_no_cw',
    'cw_general',
    'no_cw_general'
]

sample_df["cw"] = (sample_df[cw_columns] == 1).any(axis=1).astype(int)
sample_df = sample_df.drop(columns=cw_columns)
df_anno_2["cw"] = (df_anno_2[cw_columns] == 1).any(axis=1).astype(int)
df_anno_2 = df_anno_2.drop(columns=cw_columns)
```

### choose columns of interest to flatten dfs

```python
columns_of_interest = (sample_df.drop(columns = ["name", "joined_rules", "translated_rules"])).columns.to_list()
```

```python
columns_of_interest
```

```python
annotations_1 = sample_df[columns_of_interest].fillna(0).values.flatten().astype(float)
annotations_2 = df_anno_2[columns_of_interest].fillna(0).values.flatten().astype(float)
```

### calculate cohens kappa for all annotations and each column

```python
kappa = cohen_kappa_score(annotations_1, annotations_2)
print("Cohen's Kappa:", kappa)
```

```python
for col in columns_of_interest:
    y1 = sample_df[col].fillna(0).values.flatten().astype(float)
    y2 = df_anno_2[col].fillna(0).values.flatten().astype(float)
    print(f"Cohens Kappa for column {col}: {cohen_kappa_score(y1, y2)}")
```
