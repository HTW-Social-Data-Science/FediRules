```python
import os
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

from sklearn.metrics import classification_report
from sklearn.metrics import f1_score

pd.set_option('display.max_colwidth', 255)
```

```python
DATA_DIR = 'predictions_per_cat'
```

```python
pattern = r"nocw|^\.$" # remove rows with pattern
df_f1 = []
y_true_all = []
y_pred_all = []

for filename in os.listdir(DATA_DIR):
    if not filename.endswith(".parquet"):
        continue
    filepath = os.path.join(DATA_DIR, filename)
    cat_name = filename.split("predictions-rules-gpt_5_mini_")[1].split(".parquet")[0]
    df = pd.read_parquet(filepath)

    mask = df["translated_rules"].str.contains(pattern, regex=True, na=False)
    df_filtered = df[~mask]

    print(cat_name)
    print(classification_report(df[cat_name], df[f'gpt_5_mini_{cat_name}']))
    print('__________________________________________________________')

    
    y_true_all.extend(df_filtered[cat_name])
    y_pred_all.extend(df_filtered[f'gpt_5_mini_{cat_name}'])

    report = classification_report(df_filtered[cat_name], df_filtered[f'gpt_5_mini_{cat_name}'], output_dict = True)
    f1_1 = report['1']['f1-score']
    df_f1.append({"category": cat_name, "F1_GPT-5-mini_experiment": round(f1_1 * 100, 2)})

df_f1 = pd.DataFrame(df_f1)
df_f1 = df_f1.sort_values("category").reset_index(drop=True)

    
micro_f1 = round(f1_score(y_true_all, y_pred_all, average='binary') * 100, 2)
macro_f1 = round(f1_score(y_true_all, y_pred_all, average='binary') * 100, 2)
print("Micro F1 over all categories:", micro_f1)
print("Macro F1 over all categories:", macro_f1)
print(df_f1)
```
