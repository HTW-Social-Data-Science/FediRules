# calculate aggregated share for table 1

```python
AGGREGATED_GROUPS = {
    "Harassment & Hate": [
        "gpt5-hate_speech",
        "gpt5-harassment",
        "gpt5-dont_be_dick",
        "gpt5-nazis",
        "gpt5-incitement_violence",
    ],

    "Illegal & Harmful Activities": [
        "gpt5-illegal",
        "gpt5-CSAM",
        "gpt5-animal_cruelty",
        "gpt5-copyright_piracy_credit",
        "gpt5-scrape_model_training",
    ],

    "Depiction of Potentially Illegal Content": [
        "gpt5-adult_content",
        "gpt5-nsfw",
        "gpt5-violence",
    ],

    "Content Warnings": [
        "gpt5-adult_content_cw",
        "gpt5-nsfw_cw",
        "gpt5-sensitive_cw",
        "gpt5-violence_cw",
        "gpt5-politics_cw",
        "gpt5-spoilers_cw",
        "gpt5-cw_general",
    ],

    "Politics & Misinformation": [
        "gpt5-crypto",
        "gpt5-misinfo",
        "gpt5-politics",
        "gpt5-religion",
    ],

    "Platform Integrity": [
        "gpt5-spam",
        "gpt5-advertising",
        "gpt5-impersonation",
        "gpt5-automated_tools_bots",
    ],

    "Governance & Moderation": [
        "gpt5-codes_conduct",
        "gpt5-consequences_moderation",
        "gpt5-registration_requirements",
        "gpt5-private_no_reg",
    ],

    "User Norms & Guidance": [
        "gpt5-be_nice",
        "gpt5-block_report_mute",
        "gpt5-language_specific",
    ],
}
```

```python
category_shares = full_rules_df[rule_columns].mean()
aggregated_shares = {}

for group, cols in AGGREGATED_GROUPS.items():
    valid_cols = [c for c in cols if c in category_shares.index]
    aggregated_shares[group] = category_shares[valid_cols].sum()

aggregated_shares = (
    pd.Series(aggregated_shares)
    .sort_values(ascending=False)
)

aggregated_shares_percent = (aggregated_shares * 100).round(0)
print(aggregated_shares_percent)
```
