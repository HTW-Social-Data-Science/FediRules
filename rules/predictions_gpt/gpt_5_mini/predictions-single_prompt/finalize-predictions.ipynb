```python
import os
import pandas as pd
```

```python
DATA_DIR = 'partial_results'
```

```python
df_list = []
for file_name in os.listdir(DATA_DIR):
    file_path = os.path.join(DATA_DIR, file_name)
    df = pd.read_parquet(file_path)
    print(df.columns)
    print(df.shape)
    df_list.append(df)

combined_df = pd.concat(df_list, ignore_index=True)
combined_df.to_parquet('final_rules_predictions.parquet')
```
